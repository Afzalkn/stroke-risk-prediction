```python
# Core libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Display settings
pd.set_option('display.max_columns', None)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

print("✅ Libraries imported successfully!")
```

```python
# Load the dataset
df = pd.read_csv('data/healthcare-dataset-stroke-data.csv')

print(f"Dataset shape: {df.shape}")
print(f"\nColumns: {df.columns.tolist()}")
print("\n" + "="*60)
print("First few rows:")
print("="*60)
df.head()
```

```python
# Get info about the dataset
print("Dataset Information:")
print("="*60)
df.info()
```

```python
# Numerical summary
print("Statistical Summary:")
print("="*60)
df.describe()
```

```python
# Check missing values
print("Missing Values:")
print("="*60)
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
print(missing_df[missing_df['Missing Count'] > 0])

if missing_df['Missing Count'].sum() == 0:
    print("\n✅ No missing values except for BMI which we'll handle later")
else:
    print(missing_df)
```

```python
# Check target distribution
print("Stroke Distribution:")
print("="*60)
print(df['stroke'].value_counts())
print(f"\nClass Distribution (%):")
print(df['stroke'].value_counts(normalize=True).round(4) * 100)

# Visualize
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# Bar plot
df['stroke'].value_counts().plot(kind='bar', ax=ax[0], color=['#2ecc71', '#e74c3c'])
ax[0].set_title('Stroke Distribution (Count)', fontsize=14, fontweight='bold')
ax[0].set_xlabel('Stroke (0=No, 1=Yes)')
ax[0].set_ylabel('Count')
ax[0].set_xticklabels(['No Stroke', 'Stroke'], rotation=0)

# Pie chart
df['stroke'].value_counts().plot(kind='pie', ax=ax[1], autopct='%1.1f%%', 
                                    colors=['#2ecc71', '#e74c3c'],
                                    labels=['No Stroke', 'Stroke'])
ax[1].set_title('Stroke Distribution (%)', fontsize=14, fontweight='bold')
ax[1].set_ylabel('')

plt.tight_layout()
plt.savefig('reports/figures/01_target_distribution.png', dpi=100, bbox_inches='tight')
plt.show()

print("\n⚠️  KEY INSIGHT: Severe class imbalance — only ~5% positive cases!")
```

```python
# Distribution of numerical features
numerical_cols = ['age', 'avg_glucose_level', 'bmi']

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for i, col in enumerate(numerical_cols):
    sns.histplot(data=df, x=col, hue='stroke', kde=True, ax=axes[i], 
                 palette={0: '#2ecc71', 1: '#e74c3c'})
    axes[i].set_title(f'{col.upper()} Distribution by Stroke', fontweight='bold')

plt.tight_layout()
plt.savefig('reports/figures/02_numerical_distributions.png', dpi=100, bbox_inches='tight')
plt.show()
```

```python
# Categorical features analysis
categorical_cols = ['gender', 'hypertension', 'heart_disease', 'ever_married',
                     'work_type', 'Residence_type', 'smoking_status']

fig, axes = plt.subplots(3, 3, figsize=(18, 14))
axes = axes.flatten()

for i, col in enumerate(categorical_cols):
    cross_tab = pd.crosstab(df[col], df['stroke'], normalize='index') * 100
    cross_tab.plot(kind='bar', ax=axes[i], color=['#2ecc71', '#e74c3c'])
    axes[i].set_title(f'Stroke Rate by {col}', fontweight='bold')
    axes[i].set_ylabel('Percentage (%)')
    axes[i].legend(['No Stroke', 'Stroke'])
    axes[i].set_xticklabels(axes[i].get_xticklabels(), rotation=45)

# Hide unused subplots
for j in range(len(categorical_cols), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.savefig('reports/figures/03_categorical_analysis.png', dpi=100, bbox_inches='tight')
plt.show()
```

```python
# Age vs Stroke deep dive
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Boxplot
sns.boxplot(data=df, x='stroke', y='age', ax=axes[0], hue='stroke', legend=False,
            palette=['#2ecc71', '#e74c3c'])
axes[0].set_title('Age vs Stroke', fontweight='bold')
axes[0].set_xticklabels(['No Stroke', 'Stroke'])

# Stroke rate by age group
df['age_group'] = pd.cut(df['age'], bins=[0, 30, 45, 60, 75, 100],
                         labels=['<30', '30-45', '45-60', '60-75', '75+'])

age_stroke = df.groupby('age_group', observed=False)['stroke'].mean() * 100
age_stroke.plot(kind='bar', ax=axes[1], color='#e74c3c')
axes[1].set_title('Stroke Rate by Age Group (%)', fontweight='bold')
axes[1].set_ylabel('Stroke Rate (%)')
axes[1].set_xlabel('Age Group')
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=45)

plt.tight_layout()
plt.savefig('reports/figures/04_age_analysis.png', dpi=100, bbox_inches='tight')
plt.show()

print("📊 KEY INSIGHT: Stroke risk increases dramatically after age 60!")
```

```python
# Correlation heatmap (numerical features only)
plt.figure(figsize=(10, 8))
numerical_df = df.select_dtypes(include=['int64', 'float64'])
correlation = numerical_df.corr()

mask = np.triu(np.ones_like(correlation, dtype=bool))
sns.heatmap(correlation, annot=True, cmap='coolwarm', center=0, 
            mask=mask, fmt='.2f', square=True, linewidths=1)
plt.title('Correlation Matrix', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('reports/figures/05_correlation.png', dpi=100, bbox_inches='tight')
plt.show()
```
